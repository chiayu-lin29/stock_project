## 1. Clean raw data

```python
import pandas as pd
df_info = pd.read_csv("all_stock.csv")
```

```python
df_info.info()
```

```python
df_info = df_info.loc[:, ~df_info.columns.str.contains(r"^Unnamed")]
df_info.columns = df_info.columns.str.strip()
```

```python
df_info
```

```python
dup_id = df_info.groupby('stock_id')['stock_name'].nunique()
dup_id = dup_id[dup_id > 1]  # 出現多於1個名稱的ID
problem_id = df_info[df_info['stock_id'].isin(dup_id.index)]
print(problem_id)
```

```python
df = df_info.copy()
```

## preprocessing

```python
df['stock_id'] = df['stock_id'].astype(str).str.strip()
df['stock_name'] = df['stock_name'].astype(str).str.strip()
df['list_type'] = df['list_type'].astype(str).str.strip()
df['list_date'] = pd.to_datetime(df['list_date'], errors='coerce')

df = df.drop_duplicates()
```

```python
df
```

```python
id_multi_names = (df.groupby('stock_id')['stock_name']
                    .nunique().sort_values(ascending=False))
id_multi_names = id_multi_names[id_multi_names > 1]

# 列出詳細行
rows_id_multi = df[df['stock_id'].isin(id_multi_names.index)]\
                .sort_values(['stock_id','list_date','stock_name'])
```

```python
rows_id_multi
```

```python
df_sorted = df.sort_values(['stock_id', 'list_date'])
df_latest = df_sorted.drop_duplicates(subset='stock_id', keep='last')
```

```python
df_latest
```

```python
# 1) 一個名稱對到多個代號的名稱清單
dup_names = (df_latest.groupby('stock_name')['stock_id']
            .nunique().sort_values(ascending=False))
dup_names = dup_names[dup_names > 1]               # 只留 >1 的
print("重複名稱數量：", dup_names.shape[0])
print(dup_names)                             

# 2) 列出這些名稱對應的所有資料列（看清楚是哪幾個代號）
diff_rows = (df_latest[df_latest['stock_name'].isin(dup_names.index)]
             .sort_values(['stock_name','stock_id']))
diff_rows[['stock_name','stock_id','industry_category','list_type','list_date']].head(50)
```
